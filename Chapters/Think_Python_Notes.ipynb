## Debugging 

There are three different types of error 
- **Syntax Error**: means that the structure of the program is wrong. If you encounter a syntax error, your program will not run
- **Runtime Error**: an error that occurs during the runtime of a program where the code terminates and returns an error. Also called an **exception** because it indicates that somthing exceptional has happened
- **Semantic Error**: means that yoru program ran without an error message but it doesn't do what you wanted it to do

## Functions

You can create your own **functions** using the `def` keyword and then `:`. Note that when your definition of the function ends is when the indentation (four spaces or [tab]) ends.\
\
**Basic format**

````python
def my_function():
    print("This is my function")

my_function()
````

You can provide arguments to your function that defines variables that will only be excessable only in the environment of the function

````python
def square_area(side):
    area = side ** 2
    print(area)

square_area(5)
````

````python
# Calling the variable defined inside the function will give an error
print(area)
````

Function can do multiple things before it terminates, but they can also return a value using the `return` keyword. Note once you use `return` all other lines after will not be executed so you can use this to end a function early by using `return None`

````python
def square(side):
    area = side ** 2
    return area

area = square(5)
print(area)
````

When calling a function the order in which you put the arguments matters, however, if you dont want to think about it you can just include the argument names in the argument

````python
def fun1(a, b):
    return a - b

print(fun1(10, 5))
print(fun1(5, 10))

#using the names of a and b
print(fun1(b=5, a=10))
````

One can also define default values when defining a function that will auto set the local variables to that if no other value is passed in

````python
def fun2(a=2, b=2):
    return a - b

print(fun2(10))
print(fun2(b=4))
````

## `if` Statements

The basic function of `if` statement is \

>if [condition] :
> > block of code

Where the block  of code (defined by the indent) is a only executed if the if condition returns true

`if` statements have an associated keyword `pass` that will make your if statment do nothing. Useful because you cant typically have an empty if statment

## `else` Clause

You can extend extend if statements with the `else` and `elif` keywords

````python
x = 5

if x > 10:
    print('large')
elif x > 5:
    print('medium')
else :
    print('small')
````

While you can have nested conditionsals, its typically recommended against because it can become difficult to read and keep track of. Instead (when applicable) it is recommended to include all the conditions in the if statement using `and` and `or` keywords.

````python
if 1 < 2 and 3 < 4:
    print('both are true')
````
