# Introduction to Python

### Python basics

python scripts are denoted by the file type `.py`

square bracket indexing has the following format `[start:end]` where the end value is not included

```python
list = [1, 2, 3, 4, 5]
print(list[1:3])
```

You can use the `del` command to delete elements of a list

```python
l = [1, 2, 3, 4, 5]
del l[0]
print(l)
```

When you set a list equal to another list you don't create a copy, you merely refrence the original list (in python in general variables point to data instead of storing in a new location) so you can use one of the following to create a copy of an existing list

```python
# Pointing to an old list
list_1 = [1, 2, 3, 4, 5]
list_2 = [list_1, 1]
print(list_2)
list_1[0] = 0
print(list_2)
```

```python
# Creating a copy of an old list using slicing
list_1 = [1, 2, 3, 4, 5]
list_2 = list_1[:]
list_1[0] = 0
print(list_2)

# Creating a copy using the list() function
list_1 = [1, 2, 3, 4, 5]
list_2 = list(list_1)
list_1[0] = 0
print(list_2)
```

You can amend list using `+` followed by other elments enclosed by brackets

```python
l = [1, 2, 3, 4, 5]
l = l + [6]
print(l)
```

The `round(input, digts after decimal)` function rounds the funciton to n digits after the decimal, if no second value is passed in it will auto to 0

```python
print(round(3.14159, 2))
print(round(3.14159))
```

`help(function)` : tells you what a function does

### Installing Python Packages
To install python packages you have to do the following:\
1. `get-pip.py`\
Then in the terminal
2. `python3 get-pip.py`
3. `pip3 install [Package Name]`

### Importing Packages in python
Once a package is installed (look as section above) you can import it into your working directory to be used. Most commonly thats done by\
`import [package name]` \
from there you can call the functions or data types in the following way\
`ar = numpy.array([...])`\
`ar = numpy.gradient(...)`\
However, for longer package names it is much easier to use an abbreviation\
`import numpy as np`\
so now when you call the package you can use\
`ar = np.array([...])`

### Importing specific functions
Insetad of important full packages, if you want only specific functions you can use the following strcture\
`from [package] import [function]`
