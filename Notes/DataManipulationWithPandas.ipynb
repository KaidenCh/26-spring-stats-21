# Transforming Data Frames

### Introduction to Pandas

- Every value inside a column has its own data type
- different columns can contain different data types
- Pandas has different ways to quickly explore data sets
    - `df.head()` : Returns the frist few rows of a data frame
    - `df.info()` : returns the names of the columns and what type of data they contain, and if they have missing values
    - `df.shape` : returns a tupe of the number of rows and number of columns
    - `df.describe()` : computes summary statistics for each of the numerical columns
        - count : # of non missing values in each column
        - mean
        - std
        - min
        - IQR
        - max
- you can convert data frames to numpy arrays rather easily with the `df.to_numpy()` method
- you can edit the names of columns and rows of a data frame using the **attributes**
    - `df.columns`
    - `df.index` (for rows)


- the method `df.sort_values("col name")` will sort the data based on the values in that column
    - adding `("column", ascending=False)` will sort in the reverse order
    - you can add multiple columns to sort by starting by sorting the data by the first data then the second, etc if two rows have the same data value for the first column being sorted `(["col_1", "col_2"])`
    - you can choose which direction each list is sorted in by passing in a list of booleans `(["col_1", "col_2"], ascending=[True, False])
- You can subset single columns by doing `df["column_name"]`
- you can subset multiple columns with two pairs of square brackets (passing in a list) `df[["col_1", "col_2"]]`
- subsetting rows
    - most common way to subset rows is using a condition statement like `df[df["col_name"] > value]`
    - you can combine logical statements using the `&` `|`, or `and` and `or` keywords. Remember you have to add parentheses around each of the conditions
    - for catagorical variables its most common (and easy) to use the `df['col'].isin(["word 1", "word 2"])` method where you pass in a list of keywords 





- You can create new columns in a data frame by indexing a new column name that is yet to exist `df["new_col"] = ...`

# Aggregating DataFrames

- `df["col"].mean()` : method that gets the mean of the column
- you also can use the following methods on selected columns
    - .median()
    - .mode()
    - .min()
    - .max()
    - .var()
    - .std()
    - .sum()
    - .quantile()
- the `.agg(function_name)` method allows you to calculate custom summary statistics, that calls the function on the column
    - you can also use it on multiple columns at once `df[['col_1', 'col_2']].agg(func)`
    - you can pass multiple functions into `.agg()` but you have to enclose them in square brackets
- cumulative statistics : methods available to pandas that computer the cumulative statistics are
    - .`cumsum()` : returns a list of values, with the last data point being the full sum of all the data values
    - `.cummax()`
    - `.commax()`
    - `.commin()`
    - `.comprod()`


-  `df.drop_duplicates(subset="column_name")` : removes all rows that have duplicate entries in the column
    - if you want to specify multiple columns for a value to be unque in you can pass in a list `df.drop_duplicates(subset=["col_1", ...])`
- `df["col"].value_counts(sort = False, normalize=False)` :  counts the number of the different catagorical variables
    - if `sort = True` it will sort from most to least catagorical variables. If no
    - if `normalize=True` then you will turn the counts into proportions

- `df.groupby("col_1")["col_2"].method()` : will do the method on column 2 by group its retsults based on the different values of column 1
    - you can pass in multiple columns to either argument by passing in a list

- `df.pivot_table(values="col_1", index="col_2")` : `values` is the column you want to summarize, the `index` column is the columns you want to group by 
    - by default the pivot table calculates the mean
    - elif you want to use a different summary statistic you can use the `aggfunc="function_name"` argument to change it
    - you can pass in a list of functions to the `aggfunc  ` argument
    - you can sort by multiple groups by passing in additional columns in the `columns='col_name` argument
    for data that there is no data for it returns `NaN`, but if you don't want that you can set a default missing value with `fill_value=[something]`
    - by passing in the argument `margins=True` it returns a mean of all the values in the columns of the pivot table in the last row and mean of all rows in the last column



# Silicing and Indexing DataFrame

- `df.set_index("col_name")` : changes one of the columns to the index of the dataset
    - you can set multiple columns to the index using a list of the columns `.set_index(['col_1, ...])`. This is called 'multi-level or hierarchical indexes
    - if you have multi-level indexes you to index using comma seperated tuples for example of dog breed and color `.loc[[("Labrador", "Brown"), ("Chihauhau", Tan")]]`
- `df.reset_index()` : turns the index back into a column
    - `df.reset_index(drop=True)` : completely discards the index
- why indices : it makes subsetting code much simplier and cleaner
    - allows you to index using `.loc[[indices]]` for example `df.loc[['first', 'second']]`
    - values in the index neeedn't be unique (for example you can have two entries for the same type of car)
- `df.sort_index(level=[column names], ascending=[booleans])` : if you set your own index you can sort them using this method

### Slicing and subsetting with `.loc` and `iloc`
- you can sclice DataFrames but first you need to sort the index
- after sorting you can slice data `.loc` by using `:` for example `df.loc["index_name1":"index_name2"]` which will return all indecies that are between the two values
    - unlike sorting with integers this method includes the last value as well 
    - if you try and index the inner levels (if you pass in multiple columns for index) you get a blank DataFrame **doesn't work**
    - if you do want to slice inner index levels you have to index using tuples like `df.loc[("outer_1", "inner_1"), ("outer_2", "inner_2")]
    - slicing using the indicies is kind of smart **as long as the string you pass in is long enough to make it unique it will work** for dates at least
- you can also use `.loc` to slice columns
    - example: `df.loc[:, 'col_1':'col_2']` that will return columns in between columns 1 and 2, again this DOES INCLUDE COLUMN 2
- you can slice rows and columns using the `.iloc` method
    - Example: `df.iloc[n1:n2, n3:n4]` returns all the data from rows n1 to n2 and all the columns from n3:n4
    - when slicing with `iloc` the last index is not included
- note pandas won't really return errors typically they will just return blank dfs

### Working with Pivot Tables
- you can use `.loc` and `.iloc` or anything that you would apply to a data frame can be applied to pivot tables
- you can provide the arugment `axis="index"` to `.mean()` in order to mean across rows (should be used in conjunction with a pivot table)
- you can use the `axis="columns"` for `.mean()` in the same way
- if you have a date column in the standard international format `"yyyy-mm-dd"` you can use the .dt after extracting the date column to extract months, days, or years
    - `.dt.month` gives you the column as their months
    - `.dt.year` gives you the column as their years

# Creating and Visualizing DataFrames

### Visualizing Your Data
- to visualize you have to `import matploblib.pyplot as plt`
- once you have plt, you can turn any set of data into a graph by calling `.[whatever plt graph]`
    - for example you can `df['column'].hist()` to get a histogram of your columns data
- `.hist(bins=n)` : indicates how many bars there are
- additionally to get a histogram graph from a data set you can use `data.plot(kind="bar")`
    - you can change the title using the `title="title"` argument in the `.plot()`
- for line plots you can use .plot(x="column", y="another column", kind='line')
    - you can change the orientation of the x labels by passing in the `rot=[angle in degrees]` into `.plot()`
- for scatter plots you can also you x, y, but you also have to pass in `kind="scatter"`
- you can layer different plots (as long as you don't put plt.show() in between) by running the plots one after another
    - you can use `plt.legend(['col_name_1', ...])` to show the differences between data by adding a legend
    - you can use the `alpha=float` to change the transparency of data with 0 being fully see through 1 being fully there 


### Missing Values
- missing values are represented by NaN
- `df.isna()` : gets a boolean for every value if the value is missing
    - chainging it with `.any()` like `df.isna().any()` returns if any of the indices have a `NaN` value
    - chainging with `.sum()` tells us how many `NaN`s there are `df.isna().sum()`
- you can use `df.dropna()` to remove any of the rows with missing data points, however is not ideal if you have many rows with missing data points
- `df.fillna('something')` fills all the `NaN`s with 'something'


### Creating Dictionaries
- by either having a dictionary of dictionaries or a dictionary of lists (better) you can convert to a data frame using `pd.DataFrame(dictionary_name)`
