### Saving Python Objects with JSON
While `write` only can store strings and text objects you can use JSON to save other objects like numpy arrays or dictionaries and then load them another time

```python
import json

# Let's say we have a dictionary we want to save
my_data = {
    "name": "Joe Bruin",
    "age": 100
}

# We use json.dump() to write the object to a file
with open('data.json', 'w') as file:
    json.dump(my_data, file)
```

Note that uts the same process as saving textfiles just using the `json` package and methods

```python
# We use json.load() to read the file back into a Python object
with open('data.json', 'r') as file:
    loaded_data = json.load(file)

print(type(loaded_data))
print(loaded_data)
```

- `.dump()` : saves objects
- `.load()` : open objects

If you want to save multiple objects at once you have to create an appropriate dictionary

```python
# If you want to save many objects to the same file, you must create a dictionary that contains
# all of the objects you want to save, and then write that dictionary to the file.
my_list = [1, 2, 3, 4, 5]
my_string = "This is a string."
my_number = 42

# Package everything into a single dictionary
data_to_save = {
    "my_data": my_data,
    "my_list": my_list,
    "my_string": my_string,
    "my_number": my_number
}

# Save that one dictionary to the file
with open('data.json', 'w') as file:
    # Adding indent=4 makes the file easily readable for humans!
    json.dump(data_to_save, file, indent=4)
```

```python
# when you are ready to read the file back, use json.load()
# to read the entire dictionary back into memory, and then pull out the variables you want using their keys
with open('data.json', 'r') as file:
    # This loads the entire dictionary back into memory
    loaded_data = json.load(file)

# Now you can just pull your variables back out using their keys
loaded_dict = loaded_data["my_data"]
loaded_list = loaded_data["my_list"]
loaded_string = loaded_data["my_string"]
loaded_number = loaded_data["my_number"]
```

If you want to load a lot of obects but don't want to worry about storing all of them in the memory you can use JSON lines, or `.jsonl`, in addition to `.dumps()`

```python
with open('data.jsonl', 'w') as file:
    # Write each object as a JSON string on its own line
    file.write(json.dumps(my_list) + '\n')
    file.write(json.dumps(my_string) + '\n')
    file.write(json.dumps(my_number) + '\n')
```

- **Generator Expressions**
    - similar to list comphrehension but the difference is that they are lazy
    - same format as list comprehension but you use `()` instead of `[]`
    - **Example**:
    > g = (n**2 for n in range(12))
    >
    > next(g)
    every time you return `next(g)` it will return the next value
    - calling sum(g) will force it to generate all values. Though note, will not include already generated terms
    - very useful for **big datatypes** as it lets you stop once you find a match, so you dont have to iterate over the entire list
