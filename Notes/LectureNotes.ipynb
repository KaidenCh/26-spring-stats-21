### Tuples
Tuples are like lists but their elements are immutable (cannot change them) and are contained by parenthesis instead of brackets 

```python
# EXAMPLE
t = (0, 'apple', 2, 'cat', 'dog', 5, 6) # use parentheses to create a tuple
```

You canot however compare tuples to lists

```python
t = (0,'apple',2,'cat','dog',5,6) # tuple
l = [0,'apple',2,'cat','dog',5,6] # list
t == l
```

If you ever want to change a tuple you HAVE TO REPLACE THE ENTIRE TUPLE

```python
print(t)
t = ("A",) + t[1:]
t
```

NOTE: you can just index tuples in the same way you do lists or arrays

```python
l[0]
```

Like comparing strings, tuples are compared value by value until they find a value bigger / smaller

```python
#strings connections
'abce' < 'abcd'
```

```python
#EXAMPLE : they can be bigger/smaller even if the first values are the same
print((1,2, 5000) < (1,3, 1))
(1, 2, 3) < (1, 2, 1)
```

Usecase (swapping values) : it is usually annoying swapping values because you have to define a temp varaible but instead you can use a tuple

```python
a=5
b=10
a, b = b, a
print(a, b)
```

Tuples are also useful as you can define multiple variables from the output of a function

```python
addr = "mileschen@stat.ucla.edu"
uname, domain = addr.split("@")
print(uname, domain)
```

### `*` in the argument of a function
including `*` in the definition of the arugment portion of the function gathers arguments into a tuple

```python
def printall(*args):
    print(type(args))
    print(len(args))
    print(args)

printall(1, 3.0, 5, "hi", 0 , 12, "cat", [1,2,3])
```

Addtionally including `*` in the argument when you're calling the function, you turn tuples into multiple values

```python
def my_divide(x, y):
    integer = x // y
    remainder = x % y
    return integer, remainder

t = (23, 5)
```

```python
my_divide(t)
```

```python
my_divide(*t)
```

#### The `zip()` function
interweaves two or more sequences (tuples, lists, strings, etc.) until one runes out (length of the shorter one)

```python
s = 'abcde'
t = 0, 1, 2, 4
u = "apple", "banana", "cantaloupe", "durian", "efruit"
list(zip(s, t))
```

You have to turn a zip into a list or will just return a location of the zip 

```python
zip(s,t)
```

```python
for triplet in zip(s, t, u):
    print(triplet)
```

```python
iterable1 = "abcde"
iterable2 = range(5)
iterable3 = ["apple","banana","coconut","date","elderberry"]
iterable4 = (2,3,5,7,11)
x = zip(iterable1, iterable2, iterable3, iterable4)
list(x)
```
