The full the function is `plt.hist(values, bins=None, range=None, weights=None)` \
1. he bins is how many columns there are in the historgram (defaults to 10)
2. range = the range of values you should show in the form (start, end)

### plt.clf
`plt.clf` clears the graph so you can start a new one (should be used after `plt.show()`)

`plt.xlabel('text')`\

`plt.title('text')`

\
`plt.xticks([values], [names])`\

# Dictionaries & Pandas

## Dictionaries
Dicitonaries are an iterable data structure that are initialized with `{}` and have the general format
<center>d = {key_1:value_1, key_2:value2, ...}</center>

You can index the dictionary and get a value by using the key

```python
d = {1:'a', 2:'b', 3:'c'}
d[1]
```

You can get a list of the keys or a list of the values in a adictionary using the `.keys()` and the `.values()` methods

```python
print(d.keys())
print(d.values())
```

Adding any duplicate key values just replaces the value at that key with the most recent definition

The keys and values themselves in a dictionary are imutable objects (wont change if they point to a variable) but dictinaries themselves are not imutable (you can update values at certain keys or add more key value pairs). This also means you cant use lists or other changing data types as keys.

```python
a = 1
d = {1:a, 2:2, 3:3}
print(d[1])
a = 2
print(d[1])
```

```python
d[2] = 4
print(d[2])
```

You can add new keys to a dicitonary by indexing a new key that is yet to exist

```python
print(d)
d[4] = 4
print(d)
```

You can check if a key is in a dictionary using the `in` keyword

```python
5 in d
```

You can delete key value pairs in a dictionary by using the `del()` function

```python
print(d)
del(d[4])
print(d)
```

#### Lists vs Dictionaries
- lists are more useful when you need to work with large amounts of data where order matters\
- lists are more useful when you need a lookup table

#### Dictionaries of dictionaries
Dictionaries of dictionaries are allowed and will have to be index twice (just like lists of lists)

```python
d = {1:{'a':1, 'b':2}, 2:{'c':3, 'd':4}}
print(d[1]['a'])
```

#### Indexing Data Frames
you can index data frames by using:
- indexing with `[column]` : you return the column as a series, a unique data type to pandas. Can be thought of as a 1D array that can have names associated with each entry
- indexing with `[[column1, column2, ...]]` : returns a dataframe
- slicing with `[n1:n2]` : returns the rows `n1` to `n2-1` 

#### Indexing with `.loc` method
Indexing data frames using THEIR NAMES
- you can index entire rows (reutrning a data frame) using the `.loc[[row1_name, row2_name, etc]]`
- to only get specific columns for each row you can use `.loc[[ROWS], [COLUMNS]]`
- note replacing `[ROWS]` with `:` indexes all the rows

#### Indexing with `.iloc` method
Indexes data frames using their position or INDEX
- can index the rows x column of a data frame using `.iloc[[list of integers], [list of integers]]` or through `.iloc[:, :]`

# Logic, Control Flow and Filtering

### Boolean Operators

Boolean Operators: of the form\
\
statement 1 `operator` statement 2\

- `and` :  returns true if both values are true
- `or` : reutrns true if at least 1 is true
- `not` : negates the operator so `not True == False`, also has a higher priority and will be executed first

`and`, `or`, and `not` dont work with numpy arrays so instead use the following
- `logical_and()`
- `logical_or()`
- `logical_not()`

### if, elif, else

If statmentes are of the form\
\
`if [statement]:`\
&emsp; `condition`\
\
note that the colon create an environment that only gets executed if the statement is true. The evorinment ends when the indented section (4 spaces) ends after the colon.

Furthermore, you can extend the if else statments using `elif` and `else` that will check the conditions until it reaches the last else statement\
\
`if [statement]:`\
&emsp; `condition`\
`elif [statement]:`\
&emsp; `condition`\
`elif [statement]:`\
&emsp; `condition`\
&emsp; &emsp;&emsp;$\vdots$\
`else [statement]:`\
&emsp; `condition`

### Looping Over Data Structures

#### Looping over dictionaries
when looping over dictionaries to get both the key and item you have to use the `.item()` method or else the loop wont know what to iterate over because the order of dictonaries are not fixed.

```python
world = {'afghanistan': 30, 'albania': 28, 'algeria': 40}

for key, value in world.items():
    print(f"{key}: {value}")
```

#### Looping over numpy arrays
When looping over $\mathbb R^{2+}$ arrays instead of writing multipel for loops, instead it much more efficient to use the function `np.nditer(numpy_array)`

```python
import numpy as np

np_height = np.array([1.73, 1.68, 1.71, 1.89, 1.79])
np_weight = np.array([65.4, 59.2, 63.6, 88.4, 68.7])
meas = np.array([np_height, np_weight])

for i in np.nditer(meas):
    print(i)
```

### Random Numbers

Inside numpy there is a random number generator function which can be called with\
\
`np.random.rand()`\
\
you can seed your RNG with `.seed()`\
\
`np.random.seed(whatever seed you want)`\
\
though starting with the same seed you will generate the same "random" numbers each time

If you only want to get random integers you can use the function\
\
`np.random.randint(low, high=None, size=None)`\
\
Where...
- if `high` is not provided low acts like the max and you will get values from `(0, low-  1)`
- if you pass in `(n1, n2)` you get random integers from `(n1, n2 - 1)`
