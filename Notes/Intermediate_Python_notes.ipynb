# Matplotlib

```python
### plt.scatter, plt.plot
```

`.plt.scatter` and `plt.plot` are essentially the same graph but one is a scatter plot and one is a line graph. They have the following argument:\
\
`plt.scatter(x, y, s(scale), label, c, linestyle, alpha, ...)`

### plt.hist
The full the function is `plt.hist(values, bins=None, range=None, weights=None)` \
1. he bins is how many columns there are in the historgram (defaults to 10)
2. range = the range of values you should show in the form (start, end)
3. weights = how each data point is weighted

### plt.clf
`plt.clf` clears the graph so you can start a new one (should be used after `plt.show()`)

### Labels and scales

You can add labels to your plot by adding the following functions after your initial graph setup\
`plt.xlabel('text')`\
`plt.ylabel('text')`\
`plt.title('text')`

You can adjust the scale of the x and y axis using\
\
`plt.xticks([values], [names])`\
`plt.yticks([values])`\
\
where you can merely change what ticks are on the axes or change what values are on each tick (the arrays have the be the same length)

You can adjust the scale of the x and y axis using\
`plt.xscale('log')`\
`plt.yscale('scale')`

`plt.text(x, y, 'text')`: adds text to the graph starting at the given location

`plt.grid(bool)`: turns on and off the grid (based on the ticks) on a plot
