### Regular Expression (RegEx)

To use regular expression methods you have to import the appropriate package

````python
import re
````

`re.search(patter, string)` : finds the first occurance in a string, and returns a match object that contains the results of the search and `None` if it isn't there. 

````python
# .search() Example
pattern = 'cat'
text = 'The cat is on the roof.'
if re.search(pattern, text):
    print('found')
````

`.span()` : operates on the match object and returns the start and end index of the pattern

````python
result = re.search(pattern, text)
result.span()
````

`.string` : you can use this method at the end of a match object to get the original string

Raw strings `(r"...")` are recommended for regex patterns to avoid escaping issues

`re.findall()` returns all non-overlapping matches as a list

`re.search()` returns the first match as a match object

`re.finditer()` returns an iterator of match objects

RegEx symbols and character classes:
- `|` : or
- `.` : matches any single character except newline
- `^` : anchors the match to the start of a string
- `$` : anchors the match to the end of a string
- `\\` : escape special characters
- `[abc]` matches a, b, or c
- `[a-z]` matches all lowercase
- `[0-9]` matches any digit
- `[^abc]` matches any character EXCEPT a b or c
- `(abc)` : will match the exact sequence abc
- `*` : matches 0 or more repetitions
- `+` : matches 1+ repetitions
- `?` : matches 0 or 1 repetitions
- `{n}` matches exactly n repetitions
- `{n, }` matches n or more repetitions
- `{n,m}` matches between n and m repetitions
- `\d` : matches digits
- `\w` : matches any word character (letters, digits, underscore)
- `\s` : matches any whitespace
- `\D` , `\W` , `\S` : matches non-digit, non word, and non-whitespace characters respectively

Common patterns:
- `.*` - match anything (greedy)
- `.*?` - match anything (lazy/minimal)

- `.+` - match at least one character

- `\s+` - one or more whitespace characters

- `\d{3}-\d{3}-\d{4}` - US phone number format

### RegEx Symbols

`.` : matches any single character except newline

````python
text = "cat hat bat"
matches = re.findall(r".at", text)
print(f"'.at' matches: {matches}")

text2 = "c\nat"  # newline between c and at
matches = re.findall(r".at", text2)
print(f"'.at' with newline: {matches}
````

`^` : anchors the match to the start of a string

````python
text = "python is great"
match_start = re.search(r"^python", text)
print(f"^python matches: {bool(match_start)}")

text2 = "learn python"
match_start2 = re.search(r"^python", text2)
print(f"^python in 'learn python': {bool(match_start2)}")

text3 = "apple\nbanana\ncherry"
matches = re.findall(r"^\w+", text3, re.MULTILINE)
print(f"^ with MULTILINE: {matches}")
````

`$` : anchors the match to the end of a string

````python
text = "hello world"
match_end = re.search(r"world$", text)
print(f"world$ matches: {bool(match_end)}")

text2 = "world hello"
match_end2 = re.search(r"world$", text2)
print(f"world$ in 'world hello': {bool(match_end2)}")

# With MULTILINE flag, $ also matches before newlines
text3 = "apple\nbanana\ncherry"
matches = re.findall(r"\w+$", text3, re.MULTILINE)
print(f"$ with MULTILINE: {matches}")
````

`\\` : escape special characters

````python
text = "What is 2+2? 2+2=4"
# Without escape: + is a special character (one or more repetitions)
matches_wrong = re.findall(r"2+2", text)
print(f"Without escape '2+2': {matches_wrong}")  # ['22'] (matches "22" not "2+2")

# With escape: treat + as literal character
matches_correct = re.findall(r"2\+2", text)
print(f"With escape '2\\+2': {matches_correct}")  # ['2+2']

# Other examples
text2 = "Price: $50.00"
price_match = re.search(r"\$\d+\.\d{2}", text2)
print(f"Escaped $ and .: {price_match.group() if price_match else None}")  # $50.00
````

### Character Classes

`[abc]` : matches a, b, or c

````python
text = "cat bat dog rat"
matches = re.findall(r"[abc]at", text)
print(f"[abc]at matches: {matches}")  # ['cat', 'bat'] (but not 'rat')

text2 = "a b c d e f"
matches = re.findall(r"[abc]", text2)
print(f"[abc] matches: {matches}")  # ['a', 'b', 'c']
````

`[a-z]` : matches all lowercase letters

````python
text = "Hello World 123"
matches = re.findall(r"[a-z]", text)
print(f"[a-z] matches: {matches}")

# Case-insensitive alternative
matches_ci = re.findall(r"[a-zA-Z]", text)
print(f"[a-zA-Z] matches (all letters): {matches_ci}")
````

`[0-9]` : matches any digit

````python
text = "My phone: 555-123-4567"
matches = re.findall(r"[0-9]", text)
print(f"[0-9] matches: {matches}")  # ['5', '5', '5', '1', '2', '3', '4', '5', '6', '7']

# Multiple digits
matches_multiple = re.findall(r"[0-9]+", text)
print(f"[0-9]+ matches: {matches_multiple}")  # ['555', '123', '4567']
````

`(abc)` : matches the exact sequence "abc"

````python
text = "abc abcdef abc123"
matches = re.findall(r"(abc)", text)
print(f"(abc) matches: {matches}")

# With capturing group for extraction
text2 = "Color: red, Color: blue"
match = re.search(r"Color: (\w+)", text2)
if match:
    print(f"Captured group: {match.group(1)}")

# Multiple groups
text3 = "John: 25, Jane: 30"
matches = re.findall(r"(\w+): (\d+)", text3)
print(f"Multiple groups: {matches}")

# Using group for backreferences
text4 = "hello hello world"
match_backref = re.search(r"(\w+) \1", text4)
print(f"Backreference match: {match_backref.group() if match_backref else None}")
````

### Qualifiers

`*` : matches 0 or more repetitions (greedy)

````python
# Match 'go' followed by 0 or more 'o's
text = "g go goo gooo g gg"
matches = re.findall(r"go*", text)
print(f"go* matches: {matches}")  
# ['g', 'go', 'goo', 'gooo', 'g', 'g']  (note: 'g' alone matches because 0 'o's allowed)

# Practical: Match numbers with optional decimal part
text2 = "Price: 5, 10.5, 3.14, 42"
prices = re.findall(r"\d+\.?\d*", text2)
print(f"Numbers with optional decimals: {prices}")  # ['5', '10.5', '3.14', '42']

# With HTML tags (lazy vs greedy)
text3 = "<h1>Title</h1><p>Paragraph</p>"
greedy = re.findall(r"<.*>", text3)  # Greedy - matches everything
lazy = re.findall(r"<.*?>", text3)   # Lazy with ? - matches each tag separately
print(f"Greedy .* : {greedy}")  # ['<h1>Title</h1><p>Paragraph</p>']
print(f"Lazy .*? : {lazy}")      # ['<h1>', '</h1>', '<p>', '</p>']
````

`+` : matches 1 or more repetitions

````python
text = "g go goo gooo g gg"
matches = re.findall(r"go+", text)
print(f"go+ matches: {matches}")  
# ['go', 'goo', 'gooo']  (note: 'g' alone does NOT match because need at least one 'o')

# Extract numbers from mixed string
text2 = "abc123def456ghi"
numbers = re.findall(r"\d+", text2)
print(f"Numbers: {numbers}")  # ['123', '456']

# Validate email username (at least 1 character)
emails = ["a@test.com", "@test.com", "user@test.com"]
for email in emails:
    valid = re.match(r".+@.+\..+", email)
    print(f"'{email}' valid: {bool(valid)}")  # True, False, True
````

`?` : matches 0 or 1 repetitions (optional)

````python
# Match color with optional 'u' (US vs UK spelling)
text = "color colour colo colouur"
matches = re.findall(r"colou?r", text)
print(f"colou?r matches: {matches}")  # ['color', 'colour']  (not 'colo' or 'colouur')

# Match http or https
urls = ["http://example.com", "https://secure.com", "ftp://old.com"]
for url in urls:
    match = re.match(r"https?://", url)
    print(f"'{url}' has http/https: {bool(match)}")  # True, True, False

# Optional plural
text2 = "1 apple, 2 apples"
matches = re.findall(r"apples?", text2)
print(f"apples? matches: {matches}")  # ['apple', 'apples']

# Making quantifiers lazy
text3 = "<h1>Title</h1>"
greedy = re.findall(r"<.*>", text3)
lazy = re.findall(r"<.*?>", text3)
print(f"Greedy: {greedy}")  # ['<h1>Title</h1>']
print(f"Lazy: {lazy}")      # ['<h1>', '</h1>']
````

`{n}` : matches exactly n repetitions

````python
# Match exactly 3 digits
text = "123 45 6789 12345"
matches = re.findall(r"\b\d{3}\b", text)
print(f"Exactly 3 digits: {matches}")  # ['123']  (not '45', '6789', or '12345')

# Match US zip code (5 digits)
zips = ["12345", "1234", "123456", "90210"]
for zip_code in zips:
    match = re.fullmatch(r"\d{5}", zip_code)
    print(f"'{zip_code}' is valid zip: {bool(match)}")  # True, False, False, True

# Phone number format (XXX-XXX-XXXX)
phone = "555-123-4567"
match = re.fullmatch(r"\d{3}-\d{3}-\d{4}", phone)
print(f"Phone valid: {bool(match)}")  # True
````

`{n,}` : matches n or more repetitions

````python
# Match 2 or more digits
text = "a1 b22 c333 d4444 e5"
matches = re.findall(r"\d{2,}", text)
print(f"2+ digits: {matches}")  # ['22', '333', '4444']  (not '1' or '5')

# Validate minimum password length (8+ chars)
passwords = ["pass", "password123", "weak"]
for pwd in passwords:
    match = re.search(r".{8,}", pwd)
    print(f"'{pwd}' has 8+ chars: {bool(match)}")  # False, True, False

# Match lines with at least 3 words
text2 = "Hello world\nPython regex is powerful\nShort line"
lines = text2.split('\n')
for line in lines:
    if re.search(r"\w+\s+\w+\s+\w+", line):
        print(f"Has 3+ words: '{line}'")
# Output: Has 3+ words: 'Python regex is powerful'
````

`{n,m}` : matches between n and m repetitions

````python
# Match 2-4 digit numbers
text = "1 22 333 4444 55555"
matches = re.findall(r"\b\d{2,4}\b", text)
print(f"2-4 digits: {matches}")  # ['22', '333', '4444']  (not '1' or '55555')

# Validate time (HH:MM format)
times = ["12:30", "9:05", "123:45", "12:5", "24:00"]
for time in times:
    # Hours: 0-23 (2 digits), Minutes: 0-59 (2 digits)
    match = re.fullmatch(r"([01]\d|2[0-3]):[0-5]\d", time)
    print(f"'{time}' valid time: {bool(match)}")  # True, False, False, False, True

# Hex color codes (3 or 6 hex digits)
colors = ["#FFF", "#123456", "#AB", "#ABCDEF"]
for color in colors:
    match = re.fullmatch(r"#([0-9A-F]{3}|[0-9A-F]{6})", color, re.IGNORECASE)
    print(f"'{color}' valid hex: {bool(match)}")  # True, True, False, True
````

### Special Character Escapes

`\d` : matches any digit (equivalent to [0-9])

````python
text = "Order #12345: $99.99"
digits = re.findall(r"\d", text)
print(f"All digits: {digits}")  # ['1', '2', '3', '4', '5', '9', '9', '9', '9']

# Extract all numbers
numbers = re.findall(r"\d+", text)
print(f"All numbers: {numbers}")  # ['12345', '99', '99']
````

`\w` : matches word characters [a-zA-Z0-9_]

````python
text = "Hello_World 123! @#$"
words = re.findall(r"\w+", text)
print(f"Word characters: {words}")  # ['Hello_World', '123']  (underscore included)

# Extract words from text
sentence = "The quick brown fox jumps over 3 lazy dogs!"
words_only = re.findall(r"[A-Za-z]+", sentence)  # Alternative without digits/underscore
print(f"Words only: {words_only}")  # ['The', 'quick', 'brown', 'fox', 'jumps', 'over', 'lazy', 'dogs']

# Validate variable names (Python style)
var_names = ["my_var", "2var", "var123", "_private", "var-name"]
for var in var_names:
    match = re.fullmatch(r"[A-Za-z_]\w*", var)
    print(f"'{var}' valid variable: {bool(match)}")  # True, False, True, True, False
````

`\s` : matches whitespace (space, tab, newline, etc.)

````python
text = "Hello\tWorld\nPython 3.9"
whitespace = re.findall(r"\s", text)
print(f"Whitespace chars: {repr(whitespace)}")  # ['\t', '\n', ' ']

# Split on whitespace
words = re.split(r"\s+", text)
print(f"Split on whitespace: {words}")  # ['Hello', 'World', 'Python', '3.9']

# Remove extra whitespace
messy = "Too    many    spaces"
clean = re.sub(r"\s+", " ", messy).strip()
print(f"Cleaned: '{clean}'")  # 'Too many spaces'
````

`\D` : matches non-digits (opposite of `\d`)

````python
text = "abc123def456ghi"
non_digits = re.findall(r"\D+", text)
print(f"Non-digit sequences: {non_digits}")  # ['abc', 'def', 'ghi']

# Extract letters only
mixed = "a1b2c3d4e5"
letters = re.findall(r"\D", mixed)
print(f"Letters only: {letters}")  # ['a', 'b', 'c', 'd', 'e']

# Remove all digits
text_with_digits = "Order 12345 for $99.99"
no_digits = re.sub(r"\D", "", text_with_digits)
print(f"Digits only: {no_digits}")  # '123459999'
no_letters = re.sub(r"\d", "", text_with_digits)
print(f"Letters/symbols only: {no_letters}")  # 'Order  for $.'
````

`\W` : matches non-word characters (opposite of `\w`)

````python
text = "Hello, World! 123."
non_words = re.findall(r"\W+", text)
print(f"Non-word sequences: {non_words}")  # [', ', '! ', '.']

# Count punctuation in text
sentence = "Hello, world! How are you? (I'm fine.)"
punctuation = re.findall(r"\W", sentence)
print(f"Non-word chars: {punctuation}")  
# [',', ' ', '!', ' ', ' ', ' ', '?', ' ', '(', "'", ' ', '.', ')']

# Remove all non-word characters (keeping only letters, digits, underscore)
text2 = "Hello @world! How's #Python3?"
clean = re.sub(r"\W+", " ", text2).strip()
print(f"Cleaned text: '{clean}'")  # 'Hello world How s Python3'
````

`\S` : matches non-whitespace (opposite of `\s`)

````python
text = "Hello World\nPython\t3.9"
non_space = re.findall(r"\S+", text)
print(f"Non-whitespace tokens: {non_space}")  # ['Hello', 'World', 'Python', '3.9']

# Count non-space characters
sentence = "Hello world"
non_space_count = len(re.findall(r"\S", sentence))
print(f"Non-space characters: {non_space_count}")  # 10 (H,e,l,l,o,w,o,r,l,d)

# Extract words (alternative to \w+ but includes punctuation)
text2 = "Hello, world! How are you?"
words_with_punct = re.findall(r"\S+", text2)
print(f"Tokens (includes punctuation): {words_with_punct}")  
# ['Hello,', 'world!', 'How', 'are', 'you?']
````

````python
# Email validation (simplified)
emails = ["user@example.com", "first.last@domain.co.uk", "invalid@", "@missing"]
for email in emails:
    # \w+@\w+\.\w+ is too simple, here's a better one
    match = re.fullmatch(r"[\w\.-]+@[\w\.-]+\.\w+", email)
    print(f"'{email}' valid email: {bool(match)}")  # True, True, False, False
````

````python
# Extract hashtags from social media post
post = "Loving #Python and #regex! #100DaysOfCode is awesome #123"
hashtags = re.findall(r"#\w+", post)
print(f"Hashtags: {hashtags}")  # ['#Python', '#regex', '#100DaysOfCode']
````

````python
# Date validation (YYYY-MM-DD)
dates = ["2024-01-15", "2023-13-01", "2024-02-30", "99-01-01"]
for date in dates:
    match = re.fullmatch(r"\d{4}-(0[1-9]|1[0-2])-(0[1-9]|[12]\d|3[01])", date)
    print(f"'{date}' valid date format: {bool(match)}")  # True, False, False, False
````

````python
# IP address validation (simplified)
ips = ["192.168.1.1", "256.1.1.1", "10.0.0.1", "999.999.999.999"]
for ip in ips:
    # Each octet: 0-255
    pattern = r"^(25[0-5]|2[0-4]\d|[01]?\d\d?)\.(25[0-5]|2[0-4]\d|[01]?\d\d?)\.(25[0-5]|2[0-4]\d|[01]?\d\d?)\.(25[0-5]|2[0-4]\d|[01]?\d\d?)$"
    match = re.match(pattern, ip)
    print(f"'{ip}' valid IP: {bool(match)}")  # True, False, True, False
````

````python
test_numbers = [
    "(123) 456-7890",      # Valid
    "(555) 123-4567",      # Valid
    "(12) 345-6789",       # Too few digits in area code
    "(1234) 567-8901",     # Too many digits in area code
    "123-456-7890",        # Wrong format (missing parentheses)
    "(123)456-7890",       # Missing space after area code
    "(123) 456-789",       # Too few digits in local number
    "(123) 456-78901",     # Too many digits
    "(abc) 456-7890",      # Non-digits in area code
    "(123) 456-7890 ",     # Extra space at end
]
pattern = r"^\(\d{3}\) \d{3}-\d{4}$"
for number in test_numbers:
    match = re.match(pattern, number)
    print(f"'{number}' valid phone: {bool(match)}")
````

### Aliasing

There is an important dinstinction between `is` and `==` where one just compares if the values stored is the same where the other compares if they point to the exact same value

````python
a = (1, 2)
b = a
print(a == b)
print(a is b)
````
